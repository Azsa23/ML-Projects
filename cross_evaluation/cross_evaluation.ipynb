## قرايه الملف 

```python

import pandas as pd 

df = pd.read_csv('iris.csv')
df 
#هنا معلومات عن السبله الي هي ورق الاخضر الخارجيه التي تحمي برعم الزهره 
# اعمده الدخل  طول و عرض السبل و طول و عرض البتله  وهي الورقه الملونه الجذابه للزهرة 
# اما عمود الخرج  نوع الزهره 
# setona  
# virginica
# versicolor
```

## نحول عمود الخرج الى الارقام لان النموذج لا يقبل الحروف وهنا مافي داعي نستخدم labelencoder

```python
df['Species'].replace('Iris-setosa',0, inplace = True)
df['Species'].replace('Iris-virginica',1, inplace = True)
df['Species'].replace('Iris-versicolor',2, inplace = True)
```

## راح نحدد اعمده الدخل و الخرج

```python
x = df[['SepalLengthCm','SepalWidthCm','PetalLengthCm','PetalWidthCm']]
y = df['Species']
```

```python
## الان راح نسوي نموذج و نخليه يقيم النموذج عن طريق التقييم المتقاطع
from sklearn.model_selection import cross_val_score 
import sklearn.model_selection 
# الان راح نبني النموذج
from sklearn.naive_bayes import GaussianNB 
gnb = GaussianNB ()
# الان نستخدم التقييم المتقاطع 
# نحط النموذج و الدخل و الخرج و كم تكرار السي في و وش يحسب
scores_accuracy = cross_val_score(gnb , x , y , cv= 5 , scoring = 'accuracy')
# الان نحليه يحسب المتوسط 
meanscore = scores_accuracy.mean()
# الان نخليه ينضرب بالميه عشان تطلع النسبه بعدين راح نطلب منه يجيب اول رقمين بعد الفاصله 
print('accuracy = ', round(meanscore * 100 , 2))
```
