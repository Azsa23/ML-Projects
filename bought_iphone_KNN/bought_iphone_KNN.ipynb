## قرايه الملف 

```python
import pandas as pd 

dataset = pd.read_csv('iphone11.csv')

dataset
```

 ## راحح احدد اعمده الدخل و الخرج x and y 

```python
x = dataset.iloc[:,:-1].values
# لو كانت الاعمده كثيره صعب انك تكتب كل الاعمده فتكتب 
# x = dataset.iloc[:,:-1].values
y = dataset.iloc[:,3]
# تقدر تكتب بداله y= dataset.iloc[:,3]
```

## لازم نحول عمود الجنس الى ارقام يعني نرمزها (معالجه النصوص)

```python
from sklearn.preprocessing import LabelEncoder

label_gender = LabelEncoder ()

x[:,0] = label_gender.fit_transform(x[:,0])

x
```

## نقسم البيانات الى تدريب و اختبار 

```python
from sklearn.model_selection import train_test_split

x_train , x_test , y_train , y_test = train_test_split(x , y , train_size = 0.8 , random_state = 0 )
```

## تقييس اليانات 

```python
dataset.describe()
# لو تلاحظ في فرق كبير بين الاعمده من min max 
```

```python
from sklearn.preprocessing import StandardScaler 

sc =  StandardScaler()

x_train = sc.fit_transform(x_train )
x_test = sc.transform(x_test) 
```

## الان راح نسوي نموذج KNN 

# من neighbors
# ولازم تحدد كم ياخذ من الجيران k اذا طلع عدد زوجي زود واحد 
# بعد ما تطلع k تحطه داخل النموذج 
from sklearn.neighbors import KNeighborsClassifier
import math 

k = int(math.sqrt(len(x_train)))

if k % 2 == 0 :
    k = k + 1
print(k) # 41

classifier = KNeighborsClassifier (n_neighbors = k)
classifier.fit(x_train, y_train)

## الان نبيه يتنبئ 

```python
y_pred = classifier.predict(x_test)
```

## نقيم النموذج 

```python
from sklearn import metrics 

cm = metrics.confusion_matrix(y_test, y_pred)

cm 
# نحسب accuracy 
accuracy = metrics.accuracy_score(y_test, y_pred)
print("the accuracy :", accuracy)
#نحسب precision 
precision = metrics.precision_score(y_test, y_pred)
print("the precision :", precision)
#نحسب recall
recall = metrics.recall_score(y_test, y_pred)
print("the recall :", recall)
# الدقه 69
```

## الان نعطيه البيانات وهو يتنبئ 

```python
#الحين نبي نسوي تنبؤ لبعض الاشخاص اذا بيشترون او لا 
#نبي نقيس البيانات ذي 
person1 = sc.transform([[1,21,40000]])
person2 = sc.transform([[0,41,80000]])
person3 = sc.transform([[0,41,400]])

print('male aged 21 making 40k$ will buy iphone:' , classifier.predict(person1))
#عطانا صفر يعني ما يقدر يشتري صاحب البيانات ذي
print('male aged 41 making 80k$ will buy iphone:' , classifier.predict(person2))
print('female aged 41 making 400$ will buy iphone:' , classifier.predict(person3))
```

## احصاء 

```python
# الان نبي نشوف احصائيه للشخص نسبه كم يشتري وما يشتري
print('male aged 21 making 40k$ will buy iphone:' , classifier.predict_proba(person1))
print('male aged 41 making 80k$ will buy iphone:' , classifier.predict_proba(person2))
print('female aged 41 making 400$ will buy iphone:' , classifier.predict_proba(person3))
# في شجره القرار اذ طلع 0 1 يعني مارح يشتري اذا طلع العكس بيشتري 
```
