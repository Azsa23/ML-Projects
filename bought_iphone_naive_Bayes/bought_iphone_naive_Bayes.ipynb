## نقرا الملف 

```python
import pandas as pd
dataset = pd.read_csv("iphone11.csv")
dataset
# purchase_iphone صفر يعني ما يقدر يشتري و الواحد يقدر يشتري 
```

## راحح احدد اعمده الدخل و الخرج x and y 

```python
#لان هدفنا هو اعطي النموذج العمر و الراتب وهو يشوف يقدر يشتري ولا لا 
x = dataset.iloc[:,:-1].values
# لو كانت الاعمده كثيره صعب انك تكتب كل الاعمده فتكتب 
# x = dataset.iloc[:,:-1].values
y = dataset.iloc[:,3]
# تقدر تكتب بداله y= dataset.iloc[:,3]
```

## لازم نحول عمود الجنس الى ارقام يعني نرمزها (معالجه النصوص)

```python
# عن طريق LabelEncoder داخله fit_transform
from sklearn.preprocessing import LabelEncoder

label_gender = LabelEncoder()

x[:,0] = label_gender.fit_transform(x[:,0])
x
# لو تلاحظ ان عمود الجنس صار ارقام 
```

## نقسم البيانات الى تدريب و اختبار 

```python
# راح تساعدنا train_test_split من model_selection 
from sklearn.model_selection import train_test_split

x_train , x_test , y_train , y_test = train_test_split(x , y , train_size = 0.8 , random_state = 0 )
```

## تقييس اليانات 

```python
dataset.describe()
# لو تلاحظ في فرق كبير بين الاعمده من min max 
```

```python
#  علشان نقرب قيم الاعمده ببعضها و النموذج يتوقع بشكل صحيح
# راح يساعدنا StandardScaler من preprocessing
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
x_train = sc.fit_transform(x_train)
x_test = sc.transform(x_test)
```

## الان نستخدم gaussianNB من nive_bayes

```python
from sklearn.naive_bayes import GaussianNB

classifier = GaussianNB()

classifier.fit(x_train, y_train)
```

## الان نخليه يتنبئ  


```python
y_pred = classifier.predict(x_test) 
```

## الان نقيم النموذج 

```python
from sklearn import metrics

cm = metrics.confusion_matrix(y_test, y_pred)

cm
#هاذي مصفوفه الارتباك  Tp , TN FP , FN
```

```python
# نحسب accuracy 
accuracy = metrics.accuracy_score(y_test, y_pred)
print("the accuracy :", accuracy)
#نحسب precision 
precision = metrics.precision_score(y_test, y_pred)
print("the precision :", precision)
#نحسب recall
recall = metrics.recall_score(y_test, y_pred)
print("the recall :", recall)
# نلاحظ دقه تبنؤ النموذج 68 
```

## الان نبي نحط الارقام وهو يتنبئ 

```python
#الحين نبي نسوي تنبؤ لبعض الاشخاص اذا بيشترون او لا 
#نبي نقيس البيانات ذي 
person1 = sc.transform([[1,21,40000]])
person2 = sc.transform([[1,41,80000]])
person3 = sc.transform([[0,41,400]])

print('male aged 21 making 40k$ will buy iphone:' , classifier.predict(person1))
#عطانا صفر يعني ما يقدر يشتري صاحب البيانات ذي
```

```python
print('male aged 41 making 80k$ will buy iphone:' , classifier.predict(person2))
print('female aged 41 making 400$ will buy iphone:' , classifier.predict(person3))
```
