## قرايه الملف 

```python
import pandas as pd 

df = pd.read_csv('iphone11.csv')

df 
```

## راح نحدد بيانات الدخل و الخرج 

```python
x = df.iloc[:,:-1].values
y = df.iloc[:,3]
```

## الان نبي نحول عمود الجنس الئ ارقام 

```python
from sklearn.preprocessing import LabelEncoder

labelencoder_gender = LabelEncoder()

x[: , 0] = labelencoder_gender.fit_transform( x[:, 0] )
```

## الان نحتاج تقييس البيانات 

```python
from sklearn.preprocessing import StandardScaler 

sc = StandardScaler()
x = sc.fit_transform(x)
# لاحظ مارح نقسم بالطريقه العشوائي راح نسوي بالتقييم المتقاطع 
```

```python
from sklearn.model_selection import cross_val_score

# راح نسوي داله تقيم النماذج عن طريق التقيم المتقاطع
# يكتب النموذج و الدخل و الخرج  و cv 
# ثم يطلع قائمه تقييم النماذج الدقه 
def cv_comparison_classification( models , x , y ,cv):
    # نسوي اطار البيانات 
    cv_df = pd.DataFrame()
    # نسوي الحلقه يطلع الدقه و المتوسط 
    for model in models :
        acc = cross_val_score(model, x ,y ,scoring = 'accuracy',cv = cv )
        acc_avg = round(acc.mean() , 3)

        # نضيفها الى اطار البيانات
        cv_df[str(model)] = [acc_avg]
        cv_df.index = ['Accuracy']

    return cv_df    
```

## الان نستدعي النماذج التصنيف

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.neighbors import KNeighborsClassifier
from sklearn import svm

mlr_d = DecisionTreeClassifier(criterion="entropy")

mlr_g = GaussianNB()

mlr_reg = LogisticRegression()

mlr_k = KNeighborsClassifier()

mlr_svc = svm.SVC(kernel="rbf")

models = [mlr_d, mlr_g, mlr_reg, mlr_k, mlr_svc]

models
```

## الان نربطه مع الداله 

```python
comp_df = cv_comparison_classification(models, x , y , 10)
comp_df 
# لاحظ طلعنا لنا الدقه للنماذج التصنيف 
```
