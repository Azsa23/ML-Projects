## نحتاج نولد البيانات عن طريق make_classification

```python
from sklearn.datasets import make_classification 
# n_samples عدد الاسطر
# n_features عدد العواميد 
# هذا العوامل تستخدم للبيانات 
x , y = make_classification(n_samples = 1000 , n_features = 20 , n_informative = 5 , n_redundant= 15 , random_state = 1 )
print(x.shape , y.shape)
```

## نقسم البيانات الى تدريب و اختبار 

```python
from sklearn.model_selection import train_test_split

x_train , x_test , y_train , y_test = train_test_split(x , y , test_size = 0.3 )
print( x_train.shape , x_test.shape , y_train.shape , y_test.shape )
```

## الان راح نختار النموذج المناسب و انا هنا اخترت سجرة القرار 

```python
# راح نقيم اعماق شجره القرار من 1 الى 20
values = [i for i in range(1,21)]
```

```python
# هنا راح نسوي قائمه لحفظ نتيجه الاختبار و قائمه لحفظ نتيجه التدريب 
train_scores , test_scores = list(), list()
```

```python
# الان راح  نلائم الشجره بعمق معين  في مجموعه بيانات تدريب ثم نقيم الشجره في كل معادله التدريب و الاختبار 
from sklearn.metrics import accuracy_score
from sklearn.tree import DecisionTreeClassifier

for i in values :
    # الان نعطي النموذج البيانات
    model = DecisionTreeClassifier(max_depth=i)
    model.fit(x_train, y_train)

    # الان نبي نحسب بيانات تدريب  ونضيفه للقائمه للي سويناه فوق 
    train_yhat = model.predict(x_train)
    train_acc = accuracy_score(y_train, train_yhat)
    train_scores.append(train_yhat)

    # الان نبي نحسب بيانات اختبار ونضيفه للقائمه للي سويناه فوق 
    test_yhat = model.predict(x_test)
    test_acc = accuracy_score(y_test,test_yhat)
    test_scores.append(test_acc)

    # الان نطبع  ثلاث قيم بعد الفاصله 
    print('>%d , train: %.3f , test : %.3f' % (i , train_acc , test_acc))

    # لو تلاحظ عمق الشجره من واحد الئ العشرين 
    # لو تلاحظ  ان تدريب  ميه ثابته او تكون متزاده 
    # لو تلاحظ ان الاختبار كان متصاعد متنازل  
    # وهذا هو يسمئ فرط التخصيص لما تكون عاليه عند تدرييب و منخفضه عند الاختبار 
```
